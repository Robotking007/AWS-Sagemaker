# Assignment 9 — AWS SageMaker ML Pipeline

Run from the repository root with the environment installed from requirements.txt.

```python
import subprocess
import sys
from pathlib import Path

def run(script, *args):
    subprocess.run([sys.executable, script, *args], check=True)

assert Path('data/train.csv').is_file()
```

## Compare two Random Forest configurations locally

```python
run('train.py', '--n-estimators', '100', '--max-depth', '4', '--model-dir', 'artifacts/rf100')
run('train.py', '--n-estimators', '200', '--max-depth', '8', '--model-dir', 'artifacts/rf200')
```

## SageMaker training and deployment

Set the lab bucket and role, then run the following cells. AWS training and hosting incur charges. Use the README for API deployment and cleanup.

```python
region = 'ap-south-1'
bucket = 'YOUR-PRIVATE-BUCKET'
role = 'arn:aws:iam::YOUR-ACCOUNT-ID:role/SageMakerAssignmentRole'
assert bucket != 'YOUR-PRIVATE-BUCKET' and 'YOUR-ACCOUNT-ID' not in role
run('pipeline.py', '--region', region, 'train', '--bucket', bucket, '--role', role)
```

```python
model_data = 's3://YOUR-PRIVATE-BUCKET/path/to/model.tar.gz'
assert 'YOUR-PRIVATE-BUCKET' not in model_data
run('pipeline.py', '--region', region, 'deploy', '--bucket', bucket, '--role', role, '--model-data', model_data)
run('pipeline.py', '--region', region, 'predict')
```
